```python
from bert_score import score
import json
import numpy as np

with open("WO_CLASSES_RESULT.json", "r") as f:
    wo_classes_result = json.load(f)

with open("WITH_CLASSES_RESULT.json", "r") as f:
    with_classes_result = json.load(f)

original_comments_wo_classes = [str(item['original_comment']['message']) for item in wo_classes_result]
simplified_comments_wo_classes = [str(item['simplified_comment']['simplified_comment']) for item in wo_classes_result]

original_comments_with_classes = [str(item['original_comment']['message']) for item in with_classes_result]
simplified_comments_with_classes = [str(item['simplified_comment']['simplified_comment']) for item in with_classes_result]
```

```python
original_comments_wo_classes = [str(item['original_comment']['message']) for item in wo_classes_result]
simplified_comments_wo_classes = [str(item['simplified_comment']['simplified_comment']) for item in wo_classes_result]

original_comments_with_classes = [str(item['original_comment']['message']) for item in with_classes_result]
simplified_comments_with_classes = [str(item['simplified_comment']['simplified_comment']) for item in with_classes_result]
```

```python

(P, R, F), hashname = score(simplified_comments_wo_classes, original_comments_wo_classes, lang="en", return_hash=True)

results_wo_classes = []
for i in range(len(original_comments_wo_classes)):
    results_wo_classes.append({
        "original_comment": original_comments_wo_classes[i],
        "simplified_comment": simplified_comments_wo_classes[i],
        "precision": P[i].item(),
        "recall": R[i].item(),
        "f1": F[i].item()
    })

average_precision_wo_classes = np.mean([result["precision"] for result in results_wo_classes])
average_recall_wo_classes = np.mean([result["recall"] for result in results_wo_classes])
average_f1_wo_classes = np.mean([result["f1"] for result in results_wo_classes])
```

```python
top_20_worst_comments_wo_classes = sorted(results_wo_classes, key=lambda x: x["f1"])[:20]
top_20_worst_comments_with_classes = sorted(results_with_classes, key=lambda x: x["f1"])[:20]
```
